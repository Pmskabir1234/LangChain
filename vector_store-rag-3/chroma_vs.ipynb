**Vector database:** A database designed to store **vectors (embeddings)** and quickly find vectors that are **semantically similar** to a query. It is optimized for similarity search rather than exact matching.

**Vector store:** The broader concept/system that **stores embeddings along with their associated data** and lets an application retrieve similar items. A vector database is one type of vector store.

**What they solve that conventional databases struggle with:** Conventional databases are excellent at **exact/structured queries** like `WHERE name = "John"` or `price < 100`. They struggle with questions based on **meaning or similarity**, such as *“find documents about employee burnout”* when those exact words may not appear.

**Example:** If you search *“How do I reset my password?”*, a conventional database mainly looks for matching words. A vector database can find documents saying *“Steps to recover access to your account”* because their **meaning is similar**, even though the wording is different.

**In short:**
Conventional DB → **“Find this exact data.”**
Vector DB/store → **“Find data that means/is similar to this.”**


```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
MODEL_NAME = 'sentence-transformers/all-MiniLM-L6-v2'
```

```python
%pip install tf-keras
```

```python
embed_model = HuggingFaceEmbeddings(model_name=MODEL_NAME)
embed_model
```

```python
text = "I am Md Saad Bin Kabir"
docs = [
    'I love to watch cricket',
    'We should understand that passion should be the profession',
    'Coding late night with caffeine is my kinda peace'
]
```

```python
text_embedding = embed_model.embed_query(text)
print("Embedded Text: ", text_embedding)
print("Embedded Text Type: ", type(text_embedding))
print("Embedded Text Len: ", len(text_embedding))
print("Text Len: ", len(text.split(" ")))
```

```python
doc_embedding = embed_model.embed_documents(docs)
print("Embedded Doc: ", doc_embedding)
print("Embedded Doc[0] Type: ", type(doc_embedding))
print("Embedded Doc Len[0]: ", len(doc_embedding[0]))
print("Doc Len[0]: ", len(docs[0]))
```

### Real World useful examples and CRUD operations

```python
from langchain.vectorstores import Chroma
from langchain.schema import Document
```

```python
%pip install chromadb
```

```python
# Create LangChain documents for IPL players

doc1 = Document(
        page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and fitness, he has led the Royal Challengers Bangalore in multiple seasons.",
        metadata={"team": "Royal Challengers Bangalore"}
    )
doc2 = Document(
        page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm demeanor and ability to play big innings under pressure.",
        metadata={"team": "Mumbai Indians"}
    )
doc3 = Document(
        page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicketkeeping, and leadership are legendary.",
        metadata={"team": "Chennai Super Kings"}
    )
doc4 = Document(
        page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his yorkers and death-over expertise.",
        metadata={"team": "Mumbai Indians"}
    )
doc5 = Document(
        page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and match-winning performances make him a key player.",
        metadata={"team": "Chennai Super Kings"}
    )
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
vector_store = Chroma(
    embedding_function=embed_model,
    persist_directory='my_chroma_db',
    collection_name='sample'
)
```

```python
vector_store.add_documents(docs)
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
# search documents
vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2
)
```

```python
# meta-data filtering
vector_store.similarity_search_with_score(
    query="",
    filter={"team": "Chennai Super Kings"}
)
```

```python
# update documents
updated_doc1 = Document(
    page_content="Virat Kohli, the former captain of Royal Challengers Bangalore (RCB), is renowned for his aggressive leadership and consistent batting performances. He holds the record for the most runs in IPL history, including multiple centuries in a single season. Despite RCB not winning an IPL title under his captaincy, Kohli's passion and fitness set a benchmark for the league. His ability to chase targets and anchor innings has made him one of the most dependable players in T20 cricket.",
    metadata={"team": "Royal Challengers Bangalore"}
)

vector_store.update_document(document_id='09a39dc6-3ba6-4ea7-927e-fdda591da5e4', document=updated_doc1)
```

```python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python
# delete document
vector_store.delete(ids=['09a39dc6-3ba6-4ea7-927e-fdda591da5e4'])
```
